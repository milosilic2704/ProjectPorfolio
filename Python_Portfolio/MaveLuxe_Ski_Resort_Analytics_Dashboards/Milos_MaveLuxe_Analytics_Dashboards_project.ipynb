# Dashboard 1: Ski Resort Slope Density

Build an Application containing a density mapbox graph that displays color intensity based on `Total slopes`. This map should be centered roughly on the US/Canada.

* Add a slider to this application that filters the resorts mapped to those with a Price below the value selected in the slider.
* Add a radio button that toggles between resorts with night skiing and without night skiing. 
* Consider making your title dynamic based on the price input




```python
from dash import dcc, html, Dash
from dash.dependencies import Output, Input
from dash.exceptions import PreventUpdate

import plotly.express as px
import pandas as pd

resorts = (
    pd.read_csv(r"C:\Users\MilosIlic\OneDrive - Valcon Business Development A S\Data Plaftom (Power BI)\Python\Interactive Dashboards with Plotly & Dash\resources\Course_Materials\Data\Ski Resorts\resorts.csv", encoding = "ISO-8859-1")
    .query("Country in ['United States', 'Canada']")
)

resorts.head()
```

```python
from dash import dcc, html, Dash
from dash.dependencies import Output, Input
from dash.exceptions import PreventUpdate
import plotly.express as px

app = Dash(__name__)

app.layout = html.Div([
    html.H1(
        id="title",
        style={
            "textAlign": "center",
            "marginTop": "20px",
            "color": "#2c3e50"
        }
    ),
    
    html.Div([
        html.Label("Maximum Ticket Price:", style={"fontWeight": "bold", "marginBottom": "10px"}),
        dcc.Slider(
            id="budget",
            min=0,
            max=150,
            step=10,
            value=90,
            marks={i: f'${i}' for i in range(0, 160, 10)},
            tooltip={"placement": "bottom", "always_visible": True}
        ),
    ], style={"padding": "20px"}),
    
    html.Div([
        html.Label("Night Skiing Filter:", style={"fontWeight": "bold", "marginBottom": "10px"}),
        dcc.RadioItems(
            id="metric_radio",
            options=[
                {"label": "Has Night Skiing", "value": "Yes"},
                {"label": "No Night Skiing", "value": "No"},
            ],
            value="Yes",
        ),
    ], style={"padding": "20px"}),
    
    dcc.Graph(id="graph", style={"height": "600px"}),
])

@app.callback(
    [
        Output("title", "children"),
        Output("graph", "figure")
    ],
    [
        Input("budget", "value"),
        Input("metric_radio", "value")
    ]
)
def update_dashboard(budget, night_skiing):
    
    filtered_resorts = resorts[resorts["Price"] <= budget]
    
    
    filtered_resorts = filtered_resorts[filtered_resorts["Nightskiing"] == night_skiing]
    
    if filtered_resorts.empty:
        raise PreventUpdate
    
    night_ski_text = "with" if night_skiing == "Yes" else "without"
    title = f"Ski Resorts {night_ski_text} Night Skiing (Max Price: ${budget})"
    
    fig = px.density_mapbox(
        filtered_resorts,
        lat="Latitude",           
        lon="Longitude",          
        z="Total slopes",         
        radius=20,                
        center={"lat": 45, "lon": -100},  
        zoom=3,
        mapbox_style="carto-positron",
        color_continuous_scale="Viridis", 
        hover_name="Resort",      
        hover_data={
            "Total slopes": True,
            "Price": True,
            "Latitude": False,
            "Longitude": False
        }
    ).update_layout(
        margin={"r": 0, "t": 0, "l": 0, "b": 0}
    )
    
    return title, fig

if __name__ == "__main__":
    app.run(debug=True, mode = "inline", port=8054)
```

# Dashboard 2: Top Resorts by Selected Metric

Build an Application that outputs a bar chart of the top 10 resorts based on the dataframe column users select in a dropdown menu. 

* Add a checklist to allow users to select whether they want to limit results to Canada, the US, or include both countries. 
* Add an interactive element that allows users to select whether they want to sort the plotted dataframe in ascending or descending order. 
* Consider making your app title dynamic based on the column selected. 


```python
from dash import dcc, html, Dash
from dash.dependencies import Output, Input
from dash.exceptions import PreventUpdate

import plotly.express as px
import pandas as pd

resorts = (
    pd.read_csv(r"C:\Users\MilosIlic\OneDrive - Valcon Business Development A S\Data Plaftom (Power BI)\Python\Interactive Dashboards with Plotly & Dash\resources\Course_Materials\Data\Ski Resorts\resorts.csv", encoding = "ISO-8859-1")
    .query("Country in ['United States', 'Canada']")
)
```

```python
from dash import dcc, html, Dash
from dash.dependencies import Output, Input
from dash.exceptions import PreventUpdate
import plotly.express as px

app = Dash(__name__)

app.layout = html.Div([
    html.H1(
        id="title",
        style={
            "textAlign": "center",
            "marginTop": "20px",
            "color": "#2c3e50"
        }
    ),
    
    html.Div([
        html.Label("Select Countries:", style={"fontWeight": "bold", "marginBottom": "10px"}),
        dcc.Checklist(
            id="country",
            options=[
                {"label": "USA", "value": "United States"},
                {"label": "Canada", "value": "Canada"},
            ],
            value=["United States", "Canada"],
            style={"marginBottom": "10px"}
        )
    ], style={"padding": "20px"}),
    
    html.Div([
        html.Label("Select Metric:", style={"fontWeight": "bold", "marginBottom": "10px"}),
        dcc.Dropdown(
            id="list_of_columns",
            options=[
                {"label": col, "value": col} 
                for col in resorts.select_dtypes(include=['number']).columns
            ],
            value="Price",
            clearable=False
        )
    ], style={"padding": "20px"}),
    
    html.Div([
        html.Label("Sort Order:", style={"fontWeight": "bold", "marginBottom": "10px"}),
        dcc.RadioItems(
            id="radio_items",
            options=[
                {"label": " Ascending (Lowest to Highest)", "value": "asc"},
                {"label": " Descending (Highest to Lowest)", "value": "desc"},
            ],
            value="desc", 
            style={"marginBottom": "10px"}
        )
    ], style={"padding": "20px"}),
    
    dcc.Graph(id="graph", style={"padding": "20px"})
    
])

@app.callback(
    [
        Output("title", "children"),
        Output("graph", "figure")
    ],
    [
        Input("country", "value"),
        Input("list_of_columns", "value"),
        Input("radio_items", "value")
    ]
)
def update_dashboard(selected_countries, selected_column, sort_order):
    
    if not selected_countries:
        selected_countries = ["United States", "Canada"]
    
    filtered_resorts = resorts[resorts["Country"].isin(selected_countries)]
    
    if filtered_resorts.empty:
        raise PreventUpdate
    
    sorted_resorts = filtered_resorts.sort_values(
        by=selected_column,
        ascending=(sort_order == "asc")
    )
    
    top_10_resorts = sorted_resorts.head(10)
    
    countries_text = " & ".join(selected_countries)
    order_text = "Lowest" if sort_order == "asc" else "Highest"
    title = f"Top 10 {countries_text} Resorts - {order_text} {selected_column}"
    
    fig = px.bar(
        top_10_resorts,
        x="Resort",             
        y=selected_column,      
        color="Country",
        color_continuous_scale="Viridis",
        hover_data={
            "Country": True,
            "Price": True,
            selected_column: True
        }
    ).update_layout(
        yaxis={'categoryorder': 'total ascending' if sort_order == 'asc' else 'total descending'},
        showlegend=True,
        height=600
    )
    
    return title, fig

if __name__ == "__main__":
    app.run(mode="inline", debug=True, port=8054)
```
